```python
import json

purchases = {}

# Чтение исходного файла и сбор словаря
with open('purchase_log.txt', encoding='utf-8') as file:
    for line in file:
        record = json.loads(line)
        user_id = record['user_id']
        category = record['category']
        purchases[user_id] = category

# Запись словаря обратно в тот же файл
with open('purchase_log.txt', 'w', encoding='utf-8') as file:
    file.write(str(purchases))
```
